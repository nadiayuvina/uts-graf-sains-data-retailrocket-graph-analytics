```python
from neo4j import GraphDatabase, RoutingControl
URI = "neo4j://127.0.0.1:7687"
AUTH = ("neo4j", "uts_graf_sains_data_2026")
```

```python
def add_friend(driver, name, friend_name):
    driver.execute_query(
        "MERGE (a:Person {name: $name}) "
        "MERGE (friend:Person {name: $friend_name}) "
        "MERGE (a)-[:KNOWS]->(friend)",
        name=name,
        friend_name=friend_name,
        database_="neo4j",
    )
```

```python
def print_friends(driver, name):
    records, _, _ = driver.execute_query(
        "MATCH (a:Person)-[:KNOWS]->(friend) WHERE a.name = $name "
        "RETURN friend.name ORDER BY friend.name",
        name=name, database_="neo4j", routing_=RoutingControl.READ,
    )

    for record in records:
        print(record["friend.name"])
        
with GraphDatabase.driver(URI, auth=AUTH) as driver:
    add_friend(driver, "Arthur", "Guinevere")
    add_friend(driver, "Arthur", "Lancelot")
    add_friend(driver, "Arthur", "Merlin")
    print_friends(driver, "Arthur")
```

b. Membuat database baru menggunakan DDL

   CREATE DATABASE database_baru;

   Perintah tersebut digunakan untuk membuat database baru dalam instance Neo4j, apabila edisi Neo4j mendukung fitur multi-database.

c. Berpindah ke database lain
   
   :use database_baru

   Perintah tersebut digunakan untuk mengganti database aktif pada antarmuka Neo4j Browser/Query.

d. Mengakses dan memanipulasi data menggunakan DML Cypher

   Setelah memilih database tujuan, data dapat diakses menggunakan:

   MATCH (n)

   RETURN n;

   Untuk menambahkan node:

   CREATE (n:Person {name: 'Arthur'});

   Perintah MATCH digunakan untuk membaca data, sedangkan CREATE digunakan untuk menambahkan node pada database yang sedang aktif.

e. Mengakses database lain melalui Python Driver

with driver.session(database="database_baru") as session:
    result = session.run("MATCH (n) RETURN n")
    for record in result:
        print(record)

Selain menggunakan session, database lain juga dapat ditentukan melalui parameter database:
